```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm
from mlxtend.plotting import plot_decision_regions
from matplotlib.pyplot import figure
from sklearn.metrics import classification_report
```

```python
df=pd.read_csv("/content/svm_data.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df.plot()
```

```python
data_x = np.array(df[df.columns[:2]]) # segregate the input points (2-D)
data_y = np.array(df['y']) # segregate the  output class
```

```python
clf = svm.LinearSVC() # created the classifier
clf.fit(data_x, data_y) # fit the data using the above classifier
```

```python
y_pred = clf.predict(data_x)

print(classification_report(data_y, y_pred))
```

```python
figure(figsize=(10, 10))

plot_decision_regions(data_x, data_y, clf=clf, legend=2)
plt.show()
```

```python
r_clf = svm.SVC() # created the RBF classifier
r_clf.fit(data_x, data_y) # fit the data using the above classifier
```

```python
ry_pred = r_clf.predict(data_x)

print(classification_report(data_y, ry_pred))
```

```python
figure(figsize=(10, 10))

plot_decision_regions(data_x, data_y, clf=r_clf, legend=2)
plt.show()
```
